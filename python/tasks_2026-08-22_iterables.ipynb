## 1. Сумма продаж за неделю

**Сложность:** легкая

**Данные:** `sales = [1200, 850, 1430, 990, 2100, 1750, 1600]`

**Описание:** напиши функцию `total_sales(sales)`, которая возвращает сумму всех продаж.

**Вход:** список чисел.

**Выход:** число.

**Пример запуска:** `total_sales(sales)`

**Пример ответа:** `9920`

```python
# решение задачи 1
def total_sales(sales):
    sumka = 0
    for i in range (len(sales)):
        sumka += sales[i]
    return sumka    


def total_sales(sales):
    return sum(sales)
```

```python
total_sales([1200, 850, 1430, 990, 2100, 1750, 1600])
```

## 2. Средний чек

**Сложность:** легкая

**Данные:** `checks = [450, 700, 1200, 300, 850]`

**Описание:** напиши функцию `average_check(checks)`, которая возвращает среднее значение. Если список пустой, верни `0`.

**Вход:** список чисел.

**Выход:** число.

**Пример запуска:** `average_check(checks)`

**Пример ответа:** `700.0`

```python

# решение задачи 2
def average_check(checks):
    avg_check = 0
    sumka = 0
    for i in range (len(checks)):
        if checks == []:
            return 0
        else:
            sumka += checks[i]
        avg_check = sumka / len(checks)   
    return avg_check

# def average_check(checks):
    # if len(checks):
        # return sum(checks) / len(checks)   
    # return 0
```

## 3. Дни с высокими продажами

**Сложность:** легкая

**Выход:** список чисел.

**Пример запуска:** `high_sales_days(sales, limit)`

**Пример ответа:** `[2100, 1750, 1600]`

```python
# решение задачи 3
def high_sales_days(sales, limit):
    spisochek = list()
    for i in range (len(sales)):
        if sales[i] > limit:
            spisochek.append(sales[i])
    return spisochek        


def high_sales_days(sales, limit):
    return [x for x in sales if x > limit]
```

## 4. Минимум и максимум температуры

**Сложность:** легкая

**Данные:** `temperatures = [18, 21, 19, 24, 23, 17, 20]`

**Описание:** напиши функцию `min_max_temp(temperatures)`, которая возвращает словарь с ключами `min` и `max`.

**Вход:** список чисел.

**Выход:** словарь.

**Пример запуска:** `min_max_temp(temperatures)`

**Пример ответа:** `{'min': 17, 'max': 24}`

```python
# решение задачи 4
def min_max_temp(temperatures):
    minimumchik = temperatures[0]
    maximumchik = 0
    for i in range (len(temperatures)):
        if temperatures[i] < minimumchik:
            minimumchik = temperatures[i]
        if temperatures[i] > maximumchik:
            maximumchik = temperatures[i]
    return minimumchik, maximumchik               
```

```python
min_max_temp([18, 21, 19, 24, 23, 17, 20])
```

```python
# решение задачи 4
def min_max_temp(temperatures):
    minimumchik = temperatures[0]
    maximumchik = 0
    slovarchik = {}
    for i in range (len(temperatures)):
        if temperatures[i] < minimumchik:
            minimumchik = temperatures[i]
        if temperatures[i] > maximumchik:
            maximumchik = temperatures[i]
        slovarchik["min"] = minimumchik
        slovarchik["max"] = maximumchik    
    return slovarchik              
```

**Вход:** строка.

**Выход:** словарь.

**Пример запуска:** `letter_counts(word)`

**Пример ответа:** `{'с': 2, 'т': 3, 'а': 2, 'и': 2, 'к': 1}`

```python
word = 'статистика'

# решение задачи 7
def letter_counts(word):
    slovarchik = {}
    for i in word:
        if i in slovarchik:
            slovarchik[i] += 1
        else:
            slovarchik[i] = 1   
    return slovarchik        
```

## 8. Самое длинное слово

**Сложность:** легкая

**Данные:** `words = ['data', 'python', 'analysis', 'loop']`

**Описание:** напиши функцию `longest_word(words)`, которая возвращает самое длинное слово.

**Вход:** список строк.

**Выход:** строка.

**Пример запуска:** `longest_word(words)`

**Пример ответа:** `'analysis'`

```python
# решение задачи 8
def longest_word(words):
    maximumchik = 0
    wordik = words[0]
    for i in range(len(words)):
        if len(words[i]) > maximumchik:
            maximumchik = len(words[i])
            wordik = words[i]
    return wordik        
        
```
