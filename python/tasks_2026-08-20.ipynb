## 8. Повтори слово

**Сложность:** легкая

**Описание:** напиши функцию `repeat_word(word, n)`, которая печатает слово `n` раз, каждое повторение с новой строки. Используй цикл `for`.

**Вход:** строка `word` и целое число `n`.

**Выход:** строки с повторениями.

**Пример запуска:** `repeat_word("Python", 3)`

**Пример ответа:** `Python`, `Python`, `Python`

```python
# решение задачи 8
def repeat_word(word, n):
    result = ""
    for i in range (1, n+1):
        if i == n:
            result += f"{word}"
        else:
            result += f"{word}, "
    return result
```

```python
repeat_word("Asya", 2)
```

## 11. Факториал

**Сложность:** средняя

**Описание:** напиши функцию `factorial(n)`, которая возвращает произведение чисел от 1 до `n`. Для `0` результат равен `1`.

**Вход:** целое число `n`, `n >= 0`.

**Выход:** целое число.

**Пример запуска:** `factorial(5)`

**Пример ответа:** `120`

```python
# решение задачи 11
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result    
```

## 12. Степень числа

**Сложность:** средняя

**Описание:** напиши функцию `power(base, exponent)`, которая возвращает `base` в степени `exponent`. Используй цикл, не оператор `**`.

**Вход:** число `base` и целое число `exponent`, `exponent >= 0`.

**Выход:** число.

**Пример запуска:** `power(2, 5)`

**Пример ответа:** `32`

```python
# решение задачи 12
def power(base, exponent):
    result = 1
    for i in range (1, exponent + 1):
        result *= base
    return result
    
```

## 14. Сумма цифр

**Сложность:** средняя

**Описание:** напиши функцию `sum_digits(n)`, которая возвращает сумму цифр числа. Можно использовать цикл по строке.

**Вход:** целое неотрицательное число `n`.

**Выход:** целое число.

**Пример запуска:** `sum_digits(5082)`

**Пример ответа:** `15`

```python
# решение задачи 14
def sum_digits(n):
    result = 0
    while n != 0:
        result += n % 10
        n = n // 10
    return result    
```

```python
sum_digits(234)
```

## 15. Количество букв в строке

**Сложность:** средняя

**Описание:** напиши функцию `count_letter(text, letter)`, которая считает, сколько раз буква встречается в строке.

**Вход:** строка `text` и строка `letter` из одного символа.

**Выход:** целое число.

**Пример запуска:** `count_letter("банан", "а")`

**Пример ответа:** `2`

```python
# решение задачи 15
def count_letter(text, letter):
    result = 0
    for i in text:
        if i == letter:
            result += 1
    return result        
        
```

## 18. Палиндром

**Сложность:** выше средней

**Описание:** напиши функцию `is_palindrome(text)`, которая возвращает `True`, если строка читается одинаково слева направо и справа налево. Пробелы учитывать не нужно: проверяем строку без пробелов.

**Вход:** строка `text`.

**Выход:** логическое значение.

**Пример запуска:** `is_palindrome("топот")`

**Пример ответа:** `True`

```python
# решение задачи 18
def is_palindrome(text):
    if text == text[::-1]:
        return True
    return False
```

## 19. Простое число

**Сложность:** сложная

**Описание:** напиши функцию `is_prime(n)`, которая возвращает `True`, если число простое. Простое число больше 1 и делится только на 1 и на себя.

**Вход:** целое число `n`.

**Выход:** логическое значение.

**Пример запуска:** `is_prime(17)`

**Пример ответа:** `True`

```python
# решение задачи 19
def is_prime(n):
    divv = True
    for i in range (2, n):
        if n % i == 0:
            divv = False
    return divv        
    
```

```python
is_prime(17)
```

## 20. Когда вклад достигнет цели

**Сложность:** сложная

**Описание:** напиши функцию `deposit_years(start, target, percent)`, которая считает, сколько полных лет нужно, чтобы сумма `start` стала не меньше `target`, если каждый год она увеличивается на `percent` процентов. Используй цикл `while`.

**Вход:** числа `start`, `target`, `percent`.

**Выход:** целое число лет.

**Пример запуска:** `deposit_years(1000, 1500, 10)`

**Пример ответа:** `5`

```python
# решение задачи 20
def deposit_years(start, target, percent):
    blyat = 0
    while start < target:
        blyat += 1
        start = start + (start * percent / 100)
    return blyat    
```

```python
deposit_years(1000, 1500, 10)
```
